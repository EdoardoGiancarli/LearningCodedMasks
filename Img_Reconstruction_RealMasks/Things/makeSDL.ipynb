```python
from astropy.io import fits

from darksun.handle import _make_column, _make_bintable

def saveFITS(record, header, save_to):
    print("# Saving data...")
    # HDU list and Primary Header
    hdu_list = fits.HDUList([])
    primary_hdu = fits.PrimaryHDU()
    hdu_list.append(primary_hdu)

    # FITSRec
    columns = [
        _make_column(key, record[key]["data"], record[key]["format"])
        for key in record.keys()
    ]
    table_hdu = _make_bintable("DATA", columns, header)
    hdu_list.append(table_hdu)

    # save data
    hdu_list.writeto(save_to, output_verify="fix+ignore")
    hdu_list.close()
    print("# Saving completed!")
```

```python
path_data = "/mnt/d/PhD_AASS/Coding/Images_fits/" + "SDL.fits"
saveFITS(data, sdl.header, path_data)

path_catalogue = "/mnt/d/PhD_AASS/Coding/Images_fits/" + "catalogue.fits"
saveFITS(catalogue, sources.header, path_catalogue)
```
