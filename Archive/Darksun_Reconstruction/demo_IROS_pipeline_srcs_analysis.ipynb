# <center> **IROS Routine for Sources Reconstruction** <center>

- In this NB, we try to define an IROS-based reconstruction of a sky-field observed by a LEM-X camera.

- The whole procedure is built on the assumption of the priori knowledge of the sources spectra in the observed sky map, and of the camera instrumental transfer function. These data are used to compute a medium instrumental spectrum for each source, which will be used after the source association from catalogue.

- In this routine, it will be possible to perform the analysis of the sky image by using a SPECIFIC energy band, modelling the sources accordingly.

- Specifically:

    * We first perform a blind IROS reconstruction of the sky image in full band. This first loop is performed to obtain estimates of each source coordinates and total counts.
    * Then, we associate each reconstructed source with a celest object by comparison with a catalogue. From this, we access the sources IDs and corresponding spectra.
    * Finally, we perform another IROS reconstruction by selecting the detected photons in the given energy band.

- For each source, the estimated detected spectrum (emitted + camera response) should allow us to compute:

    * a counts fraction estimate in the energy band (the total counts is obtained during the first IROS loop in full band).
    * the counts histogram may be used to weight (i) the components of the source model, and (ii) the parameters of the detector spatial resolution kernel in the energy band.
